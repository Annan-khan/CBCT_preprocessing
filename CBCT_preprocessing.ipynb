```python

# CBCT PREPROCESSING

import os
import nibabel as nib
import numpy as np
import torch
import torch.nn.functional as F
from tqdm import tqdm

RAW_ROOT = "MMDental"
OUT_ROOT = "MMDental_preprocessed"
TARGET_SHAPE = (224, 224, 224)

os.makedirs(OUT_ROOT, exist_ok=True)

for fid in tqdm(os.listdir(RAW_ROOT)):
    in_path = os.path.join(RAW_ROOT, fid, f"{fid}.nii.gz")
    out_path = os.path.join(OUT_ROOT, f"{fid}.pt")

    if not os.path.exists(in_path):
        continue
    if os.path.exists(out_path):
        continue

    img = nib.load(in_path).get_fdata()
    img = np.clip(img, np.percentile(img, 0.5), np.percentile(img, 99.5))
    img = (img - img.mean()) / (img.std() + 1e-6)

    img = torch.tensor(img).float().unsqueeze(0)
    img = F.interpolate(
        img.unsqueeze(0),
        size=TARGET_SHAPE,
        mode="trilinear",
        align_corners=False
    ).squeeze(0)

    torch.save(img, out_path)
```

```python
import os
import torch
import torch.nn.functional as F
from tqdm import tqdm

INPUT_ROOT = "MMDental_preprocessed"
OUTPUT_ROOT = "MMDental_preprocessed_128"
TARGET_SHAPE = (128, 128, 128)

os.makedirs(OUTPUT_ROOT, exist_ok=True)

for fname in tqdm(os.listdir(INPUT_ROOT)):
    in_path = os.path.join(INPUT_ROOT, fname)
    out_path = os.path.join(OUTPUT_ROOT, fname)

    # Load existing .pt file
    img = torch.load(in_path)  # shape: [1, 224, 224, 224]

    # Add batch dim for interpolation
    img_resized = F.interpolate(
        img.unsqueeze(0),        # [1, 1, 224,224,224]
        size=TARGET_SHAPE,       # [128,128,128]
        mode="trilinear",
        align_corners=False
    ).squeeze(0)                # remove batch dim → [1,128,128,128]

    # Save resized tensor
    torch.save(img_resized, out_path)
```
